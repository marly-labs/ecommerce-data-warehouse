Maintenant que les tables du data warehouse sont construites (fact_sales et les dimensions produits, vendeurs et commandes), je peux passer à une phase d'analyse.

L'objectif est de produire des indicateurs métier simples afin de comprendre la performance commerciale de l'e-commerce

```python
total_ca = fact_sales['price'].sum()
total_ca
```

Le chiffre d'affaires total représente la somme de toutes les ventes enregistrées dans la table de faits. Celui-ci représente la performance globale du e-commerce sur la période couverte par les données

Maintenant que nous avons le chiffres d'affaires total, on peut analyser sa répartition par catégorie de produit.

L'objectif est d'identifier les catégories les plus contributrices au CA et de mieux comprendre les segments les plus performants du e-commerce.

Pour cela, je dois combiner la table des ventes(fact_sales) avec la dimension produit(dim_product).

```python
df = fact_sales.merge(dim_product, on='product_id', how='left')
df.head()
```

```python
ca_category = df.groupby('product_category_name')['price'].sum().sort_values(ascending=False)
ca_category
```

Le CA a été agrégé par catégorie de produit afin d'identifier les segments les plus performants.

Cette analyse permet de comprendre quelles catégories génèrent le plus de revenus et peuvent représenter des axes stratégiques pour l'entreprise.

Après avoir analysé le CA par catégorie de produit, on va maintenant identifier les vendeurs les plus performants.

L'objectif étant de comprendre quels vendeurs  génèrent le plus de revenus et contribuent le plus au CA global.

```python
top_sellers = fact_sales.groupby('seller_id')['price'].sum().sort_values(ascending=False)
top_sellers.head(10)
```

Le CA a été agrégé par vendeur afin d'identifier les meilleurs contributeurs aux ventes.
Cette analyse permet de mettre en évidence les vendeurs les plus performants et potentiellment stratégiques pour la marketplace.

A présent, je vais analyser la performance logistique des commandes en étudiant les délais de livraison.
L'objectif étant de comprendre le temps réel entre la commande et la réception par client. Ainsi que d'évaluer si les livraisons respectent les délais estimés.

```python
dim_order['order_purchase_timestamp'] = pd.to_datetime(dim_order['order_purchase_timestamp'])
dim_order['order_delivered_customer_date'] = pd.to_datetime(dim_order['order_delivered_customer_date'])

#je transforme les colonnes en vraies dates, parce que si elles restent telles quelles, python ne pourra pas faire le calcul
```

```python
dim_order['delivery_time_days'] = (
    dim_order['order_delivered_customer_date'] - dim_order['order_purchase_timestamp']
).dt.days

#je tranforme le résultat en nbre de jours
```

```python
dim_order['delivery_time_days'].mean() # je calcule le délai moyen de livraison avec la viariable que j'ai créée et stockée
```

```python
avg_delivery = dim_order['delivery_time_days'].mean() # je crée et stocke la variabe avg_delivery si je dois la réutiliser plus tard, l'afficher proprement et/ou l'intégrer dans un rapport
```

Le délai de livraison correspond au temps écoulé entre la commande et la reception par client.
Le calcul montre un délai moyen d'environ 12 jours.

Cet indicateur permet d'évaluer la performance logistique globale du marketplace.

Maintenant que le délai moyen de livraison a été calculé, on peut analyser la fiabilité des délais annoncés aux clients.

L'objectif ici étant de déterminer si les commandes sont livrées en retard par rapport à la date de livraison estimée, et dans quelle proportion

```python
#je vais convertir les dates, elles sont en str et je ne peux pas calculer Timestamp avec
dim_order['order_delivered_customer_date'] = pd.to_datetime(dim_order['order_delivered_customer_date'])
dim_order['order_estimated_delivery_date'] = pd.to_datetime(dim_order['order_estimated_delivery_date'])
```

```python
# puis je crée la variable du retard et je fais le calcul entre la date de livraison réelle et celle prévue
dim_order['delay_vs_estimation'] = (
    dim_order['order_delivered_customer_date'] - dim_order['order_estimated_delivery_date']
).dt.days

dim_order['delay_vs_estimation'].head()
```

```python
#je calcule maintenant si globalement les livraisons sont en avance ou en retard
dim_order['delay_vs_estimation'].mean()
```

```python
avg_delay_vs_estimation = dim_order['delay_vs_estimation'].mean()
```

L'analye des écarts entre la date estimée et la date réelle de livraison permet  donc d'évaluer la fiabilité des délais annoncés aux clients.

Un écart négatif indique qu'une livraison est en avance, tandis qu'un écart positif indique qu'elle est en retard.

Cette métrique est essentielle pour mesurer la performance logistique de la market place.

Après avoir calculé l'écart entre la date estimée et la date réelle de livraison, on va mesuer dès à présent la proportion de commandes livrées en retard.
Cet indicateur permet d'évaluer l'impact réel des retards sur les clients et de compléter l'analyse du délai moyen de livraison.

```python
# j'identifie les commandes en retard
dim_order['delay_vs_estimation'] > 0
```

```python
# je calcule le pourcentage
(dim_order['delay_vs_estimation'] > 0 ).mean() *100
```

L'analyse des délais de livraison montre qu'une commande est livrée en moyenne en 12 jours.
De plus, seulement environ 6,6% des commandes ont été livrées après la date estimée.
Ces résultats indiquent une bonne fiabilité logistique. La majorité des commandes étant livrées dans les délais prévus ou en avance.
